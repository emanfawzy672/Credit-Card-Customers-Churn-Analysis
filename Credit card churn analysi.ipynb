```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df["Attrition_Flag"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)
plt.ylabel("")  # removes y-axis label
plt.title("Attrition Distribution")
plt.show()
```

```python
sns.histplot(
    data=df,
    x="Customer_Age",
    hue="Attrition_Flag",
    bins=5,
    multiple="dodge"
)

plt.title("Age vs Attrition")
plt.xlabel("Age")
plt.ylabel("Customer Count")
plt.show()
```

```python
sns.histplot(
    data=df,
    x="Marital_Status",
    hue="Attrition_Flag",
    bins=5,
    multiple="dodge"
)

plt.title("Distribution of Customer Marital Status by Attrition Flag")
plt.xlabel("Marital_Status")
plt.ylabel("Customer Count")
plt.show()
```

```python
sns.histplot(
    data=df,
    x="Avg_Utilization_Ratio",
    hue="Attrition_Flag",
    bins=10,
    multiple="dodge",
    stat="percent"
)


plt.title("Average Utilization Ratio by Attrition Flag")
plt.xlabel("Avg_Utilization_Ratio")
plt.ylabel("Customer Count")
plt.show()
```
